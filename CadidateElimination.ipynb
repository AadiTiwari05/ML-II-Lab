```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Day": ["D1", "D2", "D3", "D4", "D5"],
    "Outlook": ["Sunny", "Sunny", "Sunny", "Sunny", "Sunny"],
    "Temperature": ["Mild", "Mild", "Mild", "Cool", "Mild"],
    "Humidity": ["Normal", "Normal", "Normal", "High", "High"],
    "Wind": ["Weak", "Weak", "Weak", "Weak", "Weak"],
    "Play": ["Yes", "Yes", "Yes", "Yes", "Yes"]
}

df = pd.DataFrame(data)

print(df)
```

```python
def candidate_elimination(df):

    num_attributes = df.shape[1] - 1
    S = ['0'] * num_attributes


    G = [['?'] * num_attributes]

    attributes = df.columns[:-1].tolist()
    target = df.columns[-1]

    def is_consistent(hypothesis, example):
        for i in range(len(hypothesis)):
            if hypothesis[i] != '?' and hypothesis[i] != example[i]:
                return False
        return True

    def is_more_general(h1, h2):

        for i in range(len(h1)):
            if h1[i] == '0':
                return False
            if h1[i] != '?' and h1[i] != h2[i]:
                return False
            if h1[i] == '?' and h2[i] != '?' and h2[i] != '0':
                pass
            if h1[i] != '?' and h2[i] == '?':
                 return False
        return True


    for index, row in df.iterrows():
        example = row[attributes].values.tolist()
        label = row[target]

        if label == 'Yes':
            for i in range(num_attributes):
                if S[i] == '0':
                    S[i] = example[i]
                elif S[i] != example[i]:
                    S[i] = '?'


            G = [g for g in G if is_consistent(g, example)]

        else:
            new_G = []
            for g in G:
                if is_consistent(g, example):

                    for i in range(num_attributes):
                        if g[i] == '?':

                            values_for_attribute = df[attributes[i]].unique()
                            for val in values_for_attribute:
                                if val != example[i]:
                                    g_new = list(g)
                                    g_new[i] = val

                                    if is_more_general(g_new, S) and g_new not in new_G:
                                        new_G.append(g_new)
                else:

                    if g not in new_G:
                        new_G.append(g)
            G = new_G


            G = [g_hyp for g_hyp in G if not is_more_general(S, g_hyp)]



    final_G = []
    for g1 in G:
        is_redundant = False
        for g2 in G:
            if g1 != g2 and is_more_general(g2, g1):
                is_redundant = True
                break
        if not is_redundant:
            final_G.append(g1)
    G = final_G

    return S, G


S_final, G_final = candidate_elimination(df)

print("Final Specific Hypothesis (S):", S_final)
print("Final General Hypotheses (G):", G_final)
```
